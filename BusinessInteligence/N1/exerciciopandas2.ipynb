```python
import pandas as pd


dados_alunos = {
    'Nome': ['Alice', 'Bruno', 'Carlos', 'Diana', 'Eduardo'],
    'Idade': [15, 16, 14, 17, 15]
}

df_alunos = pd.DataFrame(dados_alunos)
print("=== Questão 1 ===")
print(df_alunos)
print()



lista_funcionarios = [
    {'Nome': 'Ana', 'Cargo': 'Analista'},
    {'Nome': 'Bruno', 'Cargo': 'Gerente', 'Bonus': 500}
]

df_funcionarios = pd.DataFrame(lista_funcionarios)

# A coluna 'Bonus' da funcionária Ana recebe NaN (Not a Number),
# pois o dicionário dela não continha a chave 'Bonus'.
# O Pandas preenche automaticamente com NaN os valores ausentes.

print("=== Questão 2 ===")
print(df_funcionarios)
print()

# =============================================================
# QUESTÃO 3 - DataFrame criado a partir de duas Series
# =============================================================

indices = ['p1', 'p2', 'p3']

serie_produtos = pd.Series(['Notebook', 'Mouse', 'Teclado'], index=indices)
serie_precos  = pd.Series([3500.00, 89.90, 199.90],          index=indices)

df_produtos = pd.DataFrame({'Produto': serie_produtos, 'Preço': serie_precos})

print("=== Questão 3 ===")
print(df_produtos)
print()

# =============================================================
# QUESTÃO 4 - Novas colunas calculadas
# =============================================================

df_vendas = pd.DataFrame({
    'Sede A': [10000, 15000, 8000],
    'Sede B': [12000, 9000, 11000]
})

# 1. Total das duas sedes
df_vendas['Total Vendas'] = df_vendas['Sede A'] + df_vendas['Sede B']

# 2. Imposto = 10% da Sede A
df_vendas['Imposto'] = df_vendas['Sede A'] * 0.10

print("=== Questão 4 ===")
print(df_vendas)
print()



df_clientes = pd.DataFrame({
    'Nome':     ['Carla', 'Diego', 'Eva'],
    'CPF':      ['111.222.333-44', '555.666.777-88', '999.000.111-22'],
    'Telefone': ['(11)9999-0000', '(21)8888-1111', '(31)7777-2222'],
    'Cidade':   ['São Paulo', 'Rio de Janeiro', 'Belo Horizonte']
})

# Excluindo a coluna 'CPF' com del
del df_clientes['CPF']

# Excluindo a coluna 'Telefone' com pop() e exibindo seu conteúdo
coluna_telefone = df_clientes.pop('Telefone')

print("=== Questão 5 ===")
print("Conteúdo da coluna 'Telefone' removida:")
print(coluna_telefone)
print("\nDataFrame após remoções:")
print(df_clientes)
print()



df_paises = pd.DataFrame(
    {
        'Capital':    ['Brasília', 'Buenos Aires', 'Santiago'],
        'População':  [215_000_000, 46_000_000, 19_000_000]
    },
    index=['Brasil', 'Argentina', 'Chile']
)

print("=== Questão 6 ===")
print(df_paises.loc['Brasil'])
print()



df_carros = pd.DataFrame({
    'Modelo': ['Civic', 'Corolla', 'HB20', 'Onix', 'Polo', 'Gol'],
    'Ano':    [2022, 2021, 2023, 2022, 2020, 2019],
    'Preço':  [145000, 135000, 78000, 95000, 110000, 65000]
})

# iloc usa índices baseados em 0: 4º carro = índice 3, 5º carro = índice 4
print("=== Questão 7 ===")
print(df_carros.iloc[3:5])
print()



import numpy as np

df_historico = pd.DataFrame({
    'Data':  pd.date_range('2024-01-01', periods=20, freq='D'),
    'Valor': np.random.randint(100, 500, size=20)
})

# Linhas 5 a 12 em base 1 → índices 4 a 11 em base 0
df_slice = df_historico[4:12]

print("=== Questão 8 ===")
print(df_slice)
print()



df_loja1 = pd.DataFrame({
    'Produto': ['Caneta', 'Caderno', 'Borracha'],
    'Estoque': [200, 80, 150]
})

df_loja2 = pd.DataFrame({
    'Produto': ['Caneta', 'Régua', 'Tesoura'],
    'Estoque': [120, 60, 45]
})

df_total = pd.concat([df_loja1, df_loja2], ignore_index=True)

print("=== Questão 9 ===")
print(df_total)
print()



df_salarios = pd.DataFrame({
    'Funcionário': ['Fernanda', 'Gabriel', 'Helena', 'Igor'],
    'Salário':     [5200.0, 7800.0, 4900.0, 6100.0]
})

print("=== Questão 10 ===")

# 1. Transposta
print("1. Transposta:")
print(df_salarios.T)
print()

# 2. Tipos de dados
print("2. Tipos de dados (dtypes):")
print(df_salarios.dtypes)
print()

# 3. Total de elementos
print("3. Total de elementos (size):", df_salarios.size)
print()

# 4. Últimas duas linhas
print("4. Últimas duas linhas (tail):")
print(df_salarios.tail(2))
```
